```python
import anndata
from scipy.io import mmread
```

```python
%%time
#6min
sim_qumis_transposed = mmread('data/tasic/simulations/sim_qumi_transposed.mtx')
sim_only_marker_genes_DE_qumis_transposed = mmread('data/tasic/simulations/sim_only_marker_genes_DE_qumi_transposed.mtx')
```

```python
%%time
#8min
sim_census_transposed = mmread('data/tasic/simulations/sim_census_transposed.mtx')
sim_only_marker_genes_DE_census_transposed = mmread('data/tasic/simulations/sim_only_marker_genes_DE_census_transposed.mtx')
```

```python
%%time
#3min
adata_sim = anndata.read_h5ad('data/tasic/simulations/adata_sim_tpm.h5ad')
adata_sim_only_marker_genes_DE = anndata.read_h5ad('data/tasic/simulations/adata_sim_only_marker_genes_DE_tpm.h5ad')
```

```python
%%time
adata_sim.layers['qumis'] = sim_qumis_transposed.T.tocsc()
adata_sim.layers['census'] = sim_census_transposed.T.tocsc()
adata_sim.layers['readcounts'] = adata_sim.X.astype(int)

adata_sim_only_marker_genes_DE.layers['qumis'] = sim_only_marker_genes_DE_qumis_transposed.T.tocsc()
adata_sim_only_marker_genes_DE.layers['census'] = sim_only_marker_genes_DE_census_transposed.T.tocsc()
adata_sim_only_marker_genes_DE.layers['readcounts'] = adata_sim_only_marker_genes_DE.X.astype(int)
```

```python
%%time
adata_sim.write_h5ad('data/tasic/simulations/adata_sim_qumi_census_tpm.h5ad')
adata_sim_only_marker_genes_DE.write_h5ad('data/tasic/simulations/adata_sim_only_marker_genes_DE_qumi_census_tpm.h5ad')
```
