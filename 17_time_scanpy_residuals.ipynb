```python
import numpy as np
import scanpy as sc
import anndata
sc.settings.verbosity = 2
from readcount_tools import compute_residuals

import time
```

```python
import logging
logging.basicConfig(filename='03e_runtimes_scanpy_residuals.log', level=logging.INFO,force=True)

```

```python
def scanpy_baseline(ad):
    ad.layers['logmedian'] =  sc.pp.normalize_total(ad,inplace=False)['X']
    sc.pp.log1p(ad,layer='logmedian')
```

```python
ns = [1000, 5000, 10000, 20000, 50000, 100000]
```

```python
basepath = 'data/tasic/runtime_exp/'



for n in ns:

    ad = anndata.read_h5ad(f'{basepath}adata_{n}.h5ad')
    
    logging.info(f'\n\n================ Running for {ad.shape}================ ')
    
    start_time = time.time()
    compute_residuals(ad,alpha=50,theta=100)
    logging.info(f"---Compound Pearson residuals: {time.time() - start_time} seconds ---")
    
    start_time = time.time()
    scanpy_baseline(ad)
    logging.info(f"---Scanpy normalize & log transform {time.time() - start_time} seconds ---")
    logging.info('================ ================ ================ ================ \n')

    del ad
```
